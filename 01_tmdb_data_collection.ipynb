```python
import requests
import json
from dotenv import load_dotenv
import os
```

```python
url = "https://api.themoviedb.org/3/discover/movie"
params = {
    "api_key": api_key,
    "language": "en-US",
    "sort_by": "popularity.desc",
    "page": 1
}

response = requests.get(url, params=params)
print("Status code:", response.status_code)

movies_data = response.json()
print("Number of movies on this page:", len(movies_data["results"]))
```

```python
first_movie = movies_data["results"][0]
print(first_movie)
```

```python
clean_movie = {
    "id": first_movie["id"],
    "title": first_movie["title"],
    "genre_ids": first_movie["genre_ids"],
    "release_date": first_movie["release_date"],
    "original_language": first_movie["original_language"],
    "popularity": first_movie["popularity"],
    "vote_average": first_movie["vote_average"],
    "vote_count": first_movie["vote_count"]
}

print(clean_movie)
```

```python
movies_list = []

for movie in movies_data["results"]:
    clean_movie = {
        "id": movie["id"],
        "title": movie["title"],
        "genre_ids": movie["genre_ids"],
        "release_date": movie["release_date"],
        "original_language": movie["original_language"],
        "popularity": movie["popularity"],
        "vote_average": movie["vote_average"],
        "vote_count": movie["vote_count"]
    }
    movies_list.append(clean_movie)

print("Total movies collected:", len(movies_list))
print(movies_list[0])
```

```python
import time

movies_list = []
total_pages = 150

for page in range(1, total_pages + 1):
    params = {
        "api_key": api_key,
        "language": "en-US",
        "sort_by": "popularity.desc",
        "page": page
    }
    
    response = requests.get(url, params=params)
    page_data = response.json()
    
    for movie in page_data["results"]:
        clean_movie = {
            "id": movie["id"],
            "title": movie["title"],
            "genre_ids": movie["genre_ids"],
            "release_date": movie["release_date"],
            "original_language": movie["original_language"],
            "popularity": movie["popularity"],
            "vote_average": movie["vote_average"],
            "vote_count": movie["vote_count"]
        }
        movies_list.append(clean_movie)
    
    time.sleep(0.3)
    
    if page % 25 == 0:
        print("Completed page:", page)

print("TOTAL MOVIES COLLECTED:", len(movies_list))
```

```python
tv_url = "https://api.themoviedb.org/3/discover/tv"
tv_list = []
total_pages = 150

for page in range(1, total_pages + 1):
    params = {
        "api_key": api_key,
        "language": "en-US",
        "sort_by": "popularity.desc",
        "page": page
    }
    
    response = requests.get(tv_url, params=params)
    page_data = response.json()
    
    for show in page_data["results"]:
        clean_show = {
            "id": show["id"],
            "name": show["name"],
            "genre_ids": show["genre_ids"],
            "first_air_date": show["first_air_date"],
            "original_language": show["original_language"],
            "popularity": show["popularity"],
            "vote_average": show["vote_average"],
            "vote_count": show["vote_count"]
        }
        tv_list.append(clean_show)
    
    time.sleep(0.3)
    
    if page % 25 == 0:
        print("Completed page:", page)

print("TOTAL TV SHOWS COLLECTED:", len(tv_list))
```

```python
# Movie genres
movie_genre_url = "https://api.themoviedb.org/3/genre/movie/list"
params = {"api_key": api_key, "language": "en-US"}
response = requests.get(movie_genre_url, params=params)
movie_genres = response.json()["genres"]

# TV genres
tv_genre_url = "https://api.themoviedb.org/3/genre/tv/list"
response = requests.get(tv_genre_url, params=params)
tv_genres = response.json()["genres"]

# Combine both into one master dictionary
genre_lookup = {}
for genre in movie_genres + tv_genres:
    genre_lookup[genre["id"]] = genre["name"]

print("Total unique genres in lookup:", len(genre_lookup))
print(genre_lookup)
```

```python
# Map genres for movies
for movie in movies_list:
    genre_names = []
    for gid in movie["genre_ids"]:
        name = genre_lookup.get(gid, "Unknown")
        genre_names.append(name)
    movie["genres"] = ", ".join(genre_names)

# Map genres for TV shows
for show in tv_list:
    genre_names = []
    for gid in show["genre_ids"]:
        name = genre_lookup.get(gid, "Unknown")
        genre_names.append(name)
    show["genres"] = ", ".join(genre_names)

# Quick check
print(movies_list[0]["genres"])
print(tv_list[0]["genres"])
```

```python
import pandas as pd

# Convert to DataFrames
movies_df = pd.DataFrame(movies_list)
tv_df = pd.DataFrame(tv_list)

# Save to CSV
movies_df.to_csv("tmdb_movies.csv", index=False)
tv_df.to_csv("tmdb_tvshows.csv", index=False)

# Confirmation
print("Movies saved:", movies_df.shape)
print("TV Shows saved:", tv_df.shape)
print("\nMovies columns:", list(movies_df.columns))
print("TV columns:", list(tv_df.columns))
```

```python
# Remove duplicates - keep first occurrence of each id
movies_df = movies_df.drop_duplicates(subset='id', keep='first')

# Remove blank genres
movies_df = movies_df.dropna(subset=['genres'])

# Same for TV shows
tv_df = tv_df.drop_duplicates(subset='id', keep='first')
tv_df = tv_df.dropna(subset=['genres'])

# Save cleaned final versions
movies_df.to_csv("tmdb_movies_final.csv", index=False, encoding="utf-8-sig")
tv_df.to_csv("tmdb_tvshows_final.csv", index=False, encoding="utf-8-sig")

print("=== MOVIES ===")
print("Total rows:", len(movies_df))
print("Duplicate IDs:", movies_df['id'].duplicated().sum())
print("Blank genres:", movies_df['genres'].isna().sum())

print("\n=== TV SHOWS ===")
print("Total rows:", len(tv_df))
print("Duplicate IDs:", tv_df['id'].duplicated().sum())
print("Blank genres:", tv_df['genres'].isna().sum())
```

```python
import pandas as pd
import mysql.connector

movies_df = pd.read_csv("tmdb_movies_final.csv")
tv_df = pd.read_csv("tmdb_tvshows_final.csv")

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="Myyy@1612",
    database="ott_content_strategy"
)
cursor = conn.cursor()

# Insert movies
movies_inserted = 0
for _, row in movies_df.iterrows():
    sql = """INSERT INTO movies 
             (id, title, genre_ids, release_date, original_language, 
              popularity, vote_average, vote_count, genres, content_type)
             VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s)"""
    values = (
        int(row['id']),
        str(row['title']),
        str(row['genre_ids']),
        None if pd.isna(row['release_date']) else str(row['release_date']),
        str(row['original_language']),
        float(row['popularity']),
        float(row['vote_average']),
        int(row['vote_count']),
        str(row['genres']),
        str(row['content_type'])
    )
    cursor.execute(sql, values)
    movies_inserted += 1

conn.commit()
print(f"Movies inserted: {movies_inserted}")

# Insert TV shows
tv_inserted = 0
for _, row in tv_df.iterrows():
    sql = """INSERT INTO tvshows 
             (id, title, genre_ids, release_date, original_language, 
              popularity, vote_average, vote_count, genres, content_type)
             VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s)"""
    values = (
        int(row['id']),
        str(row['title']),
        str(row['genre_ids']),
        None if pd.isna(row['release_date']) else str(row['release_date']),
        str(row['original_language']),
        float(row['popularity']),
        float(row['vote_average']),
        int(row['vote_count']),
        str(row['genres']),
        str(row['content_type'])
    )
    cursor.execute(sql, values)
    tv_inserted += 1

conn.commit()
print(f"TV Shows inserted: {tv_inserted}")

cursor.close()
conn.close()
print("Done!")
```
